## Section C — Hard (5 Questions)

### Q26. Create `course_df`; display columns, retrieve Student, select Student and Score, then add Batch.

```python
course_df = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Neha", "Kabir", "Mina"],
    "Course": ["Python", "Pandas", "Python", "NumPy", "Pandas"],
    "Score": [88, 91, 79, 85, 95]
})
print(course_df.columns)
print(course_df["Student"])
student_score_df = course_df[["Student", "Score"]]
print(student_score_df)
course_df["Batch"] = "2026"
course_df
```

### Q27. Create `travel_df`, add Code with `np.arange(5)`, then replace it with a partially indexed Series.

```python
travel_df = pd.DataFrame({
    "Destination": ["Delhi", "Agra", "Jaipur", "Mumbai", "Pune"]
})
travel_df["Code"] = np.arange(5)
print("After assigning np.arange(5):")
print(travel_df)

travel_codes = pd.Series([101, 104], index=[1, 4])
travel_df["Code"] = travel_codes
print("After assigning the partially indexed Series:")
print(travel_df)
```

### Q28. Create `school_df`, request missing House, then fill the whole House column.

```python
school_df = pd.DataFrame({
    "Student": ["Isha", "Arjun", "Tara", "Dev"],
    "Class": [6, 7, 6, 8],
    "Section": ["A", "B", "A", "C"]
})
school_requested_df = school_df.reindex(columns=["Student", "Class", "Section", "House"])
print("Before assigning House:")
print(school_requested_df)
school_requested_df["House"] = "Blue House"
school_requested_df
```

### Q29. Add a column from a partially indexed Series, then delete it.

```python
inventory_df = pd.DataFrame({
    "Item": ["Pen", "Notebook", "Marker", "Folder", "Ruler"],
    "Quantity": [20, 15, 8, 12, 10]
})
stock_locations = pd.Series(["Shelf A", "Shelf C"], index=[0, 3])
inventory_df["Location"] = stock_locations
print("After adding Location:")
print(inventory_df)
del inventory_df["Location"]
print("After deleting Location:")
print(inventory_df)
```
