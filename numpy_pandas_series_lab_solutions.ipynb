```python
# Q24. Product costs, total bill, and percentage contribution
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
product_costs = quantity * price
total_bill = product_costs.sum()
percentages = product_costs / total_bill * 100
print("Cost per product:", product_costs)
print("Overall bill:", total_bill)
print("Percentage contribution (%):", np.round(percentages, 2))
```

```python
# Q25. Row-wise and column-wise variance and standard deviation
matrix = np.array([[2, 4, 6, 8], [3, 6, 9, 12], [5, 10, 15, 20]])
print("Matrix:\n", matrix)
print("Row variances (axis=1):", np.var(matrix, axis=1))
print("Row standard deviations (axis=1):", np.std(matrix, axis=1))
print("Column variances (axis=0):", np.var(matrix, axis=0))
print("Column standard deviations (axis=0):", np.std(matrix, axis=0))
# axis=0 reduces down rows, giving one result per column; axis=1 reduces across columns, giving one result per row.
```

```python
# Q26. Matrix multiplication of 2 x 3 and 3 x 2 matrices
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[1, 2], [3, 4], [5, 6]])
print("A shape:", A.shape)
print("B shape:", B.shape)
print("A @ B:\n", A @ B)
# Element-wise multiplication is not directly possible because shapes (2, 3) and (3, 2) are not broadcast-compatible.
# Matrix multiplication works because A's columns (3) equal B's rows (3).
```

```python
# Q28. View versus independent copy
# b is a view of a; c is an independent copy.
a = np.array([5, 10, 15, 20, 25, 30])
b = a[1:5]
c = a[1:5].copy()
b[1] = 99
c[2] = 77
print("a:", a)
print("b:", b)
print("c:", c)
# Editing b changes a because b is a view. Editing c does not affect a because c is a copy.
```

```python
# Q35. Reproducible 12 x 4 marks analysis; students numbered 1-12
rng = np.random.default_rng(2026)
marks = rng.integers(20, 101, size=(12, 4))
student_averages = marks.mean(axis=1)
test_means = marks.mean(axis=0)
test_stds = marks.std(axis=0)
pass_all = np.all(marks >= 40, axis=1)
high_average_students = np.where(student_averages >= 75)[0] + 1
greatest_variation_test = int(test_stds.argmax()) + 1
print("Marks (rows = students 1-12, columns = Tests 1-4):\n", marks)
print("Student averages:", np.round(student_averages, 2))
print("Test means:", np.round(test_means, 2))
print("Test standard deviations:", np.round(test_stds, 2))
print("Students passing all four tests:", (np.where(pass_all)[0] + 1).tolist())
print("Students with average >= 75:", high_average_students.tolist())
print("Test with greatest variation: Test", greatest_variation_test)
```

```python
# Q36. Central 3 x 3 matrix, transpose, product, row sums, maximum positions
matrix = np.arange(21, 46).reshape(5, 5)
central = matrix[1:4, 1:4]       # shape (3, 3)
central_T = central.T            # shape (3, 3)
product = central @ central_T    # shape (3, 3)
row_sums = product.sum(axis=1)
maximum = product.max()
positions = np.argwhere(product == maximum)  # zero-based (row, column) positions
print("Original shape:", matrix.shape)
print("Central matrix:\n", central, "\nShape:", central.shape)
print("Transpose:\n", central_T, "\nShape:", central_T.shape)
print("Product:\n", product, "\nShape:", product.shape)
print("Row-wise sums:", row_sums)
print("Maximum product value:", maximum)
print("Positions tied at maximum (zero-based):", positions.tolist())
```

```python
# Q37. Six days x four products sales and discount analysis
quantities = np.array([
    [20, 25, 35, 18],
    [22, 20, 30, 22],
    [18, 24, 32, 25],
    [25, 27, 28, 20],
    [21, 23, 40, 19],
    [19, 29, 36, 24]
])
original_prices = np.array([150, 240, 90, 320])
daily_revenue = (quantities * original_prices).sum(axis=1)
product_quantity_totals = quantities.sum(axis=0)
product_revenue_totals = (quantities * original_prices).sum(axis=0)
print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue_totals)
print("Highest-revenue day: Day", int(daily_revenue.argmax()) + 1)
print("Best-selling product by quantity: Product", int(product_quantity_totals.argmax()) + 1)
print("Total quantity per product:", product_quantity_totals)
discounted_prices = original_prices.astype(float).copy()
discounted_prices[product_quantity_totals > 100] *= 0.90
new_total_revenue = (quantities * discounted_prices).sum()
print("Original prices (unchanged):", original_prices)
print("Discounted prices:", discounted_prices)
print("Revenue after discounts:", new_total_revenue)
```

```python
# Q38. Generate normal data, clip values to [30, 70], compare statistics
rng = np.random.default_rng(2026)
observations = rng.normal(loc=50, scale=12, size=100)
outside_count = np.count_nonzero((observations < 30) | (observations > 70))
cleaned = np.where(observations < 30, 30, np.where(observations > 70, 70, observations))
def summarize(data):
    return {
        "mean": float(np.mean(data)),
        "variance": float(np.var(data)),
        "standard deviation": float(np.std(data)),
        "minimum": float(np.min(data)),
        "maximum": float(np.max(data)),
    }
print("Values outside [30, 70]:", outside_count)
print("Original statistics:", summarize(observations))
print("Cleaned statistics:", summarize(cleaned))
# Boundary replacement (clipping) limits extremes and therefore cannot increase the range; it usually reduces variance/std.
# The mean may move either way depending on the sampled tails.
```

### Pandas Series

```python
# Q39. Three branches: ordinary addition and absent records treated as zero
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})
ordinary = branch_a + branch_b + branch_c
print("Ordinary addition (NaN if any branch is missing):\n", ordinary)
common_quarters = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
common_total = branch_a[common_quarters] + branch_b[common_quarters] + branch_c[common_quarters]
print("Quarters common to all branches:", common_quarters.tolist())
print("Combined earnings for common quarters:\n", common_total)
zero_as_absent = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("Result treating absent records as zero:\n", zero_as_absent)
# Zero filling treats unrecorded branches as contributing zero; ordinary addition keeps those totals unknown (NaN).
```

```python
# Q40. Preserve required index order, detect missing students, filter scores >= 70
scores = {"R101": 78, "R102": 65, "R104": 92, "R105": 48}
required = ["R105", "R103", "R101", "R106", "R104", "R102"]
all_scores = pd.Series(scores, index=required)
print("Scores in required order:\n", all_scores)
print("Missing student records:", all_scores[all_scores.isnull()].index.tolist())
valid_high_scores = all_scores[all_scores.notnull() & (all_scores >= 70)]
print("Valid scores >= 70:\n", valid_high_scores)
filtered_dict = valid_high_scores.to_dict()
reconstructed = pd.Series(filtered_dict)
print("Dictionary:", filtered_dict)
print("Reconstructed Series:\n", reconstructed)
print("Labelled values match:", reconstructed.equals(valid_high_scores))
# Missing records should remain missing, not be converted to zero, because no score was recorded.
```
