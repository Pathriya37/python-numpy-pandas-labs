## Topic 3: Objects, Identity (`is`) vs Equality (`==`)

### Q3.1 — Difference between `is` and `==`

`==` compares values for equality. `is` checks whether two names refer to the exact same object in memory.

### Q3.2 — Compare two lists

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

**Output:** `True`, then `False`. The lists have equal contents but are separate list objects.

### Q3.3 — Compare small and larger integers

```python
x = 100
y = 100
print(x is y)
print(x == y)

p = 1000
q = 1000
print(p is q)
print(p == q)
```

In many standard CPython runs, small integers are cached and may share identity. Identity results for larger integers can depend on the implementation and how code is executed. The equality comparisons are `True`; use `==`, not `is`, to compare numeric values.

```python
i = 0
condition_checks = 0
while True:
    condition_checks += 1
    if not (i < 10):
        break
    print(i)
    i += 1
print("Condition evaluated", condition_checks, "times")
```

## Topic 7: Functions & Passing Arguments

### Q7.1 — Implement absolute value without `abs()`

```python
def my_abs(val):
    if val < 0:
        return -val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

### Q7.2 — What happens with `my_abs("Hi")`?

```python
def my_abs(val):
    if val < 0:
        return -val
    return val

try:
    print(my_abs("Hi"))
except TypeError as error:
    print(type(error).__name__ + ":", error)
```

A `TypeError` occurs because Python cannot compare a string with the integer `0` using `<` here.

### Q7.3 — Print absolute value, but do not return it

```python
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

**Output:** `2.7` and then `None`. The function prints the absolute value but has no explicit `return`, so the assignment to `x` receives `None`.

### Q7.4 — Reassigning a local parameter

```python
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)
```

**Output:** `7`. The reassignment changes only the local parameter `val`, not the caller variable `x`.

### Q7.5 — Swap local variables

```python
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:", val1, val2)

x = 6
y = 3
swap_message(x, y)
print("Original variables:", x, y)
```

## Topic 8: Scope — Local vs Global Variables

### Q8.1 — Accessing a local parameter outside a function

```python
def my_abs(val):
    if val < 0:
        return -val
    return val

try:
    print(val)
except NameError as error:
    print(type(error).__name__ + ":", error)
```

A `NameError` occurs because `val` was only defined as a local parameter inside the function.

### Q8.2 — Read a global variable

```python
counter = 0

def show_counter():
    print(counter)

show_counter()
```

### Q8.3 — Try to modify a global without `global`

```python
counter = 0

def increment_counter_wrong():
    counter = counter + 1

try:
    increment_counter_wrong()
except UnboundLocalError as error:
    print(type(error).__name__ + ":", error)
```

`UnboundLocalError` occurs because assignment makes `counter` local to the function, but the right-hand side tries to read it before local initialization.

### Q8.4 — Correctly increment the global

```python
counter = 0

def increment_counter():
    global counter
    counter += 1

increment_counter()
print(counter)
```
