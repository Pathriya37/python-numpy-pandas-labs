### Q26. Add north and south earnings and explain alignment.

```python
north = Series({"Jan": 40, "Feb": 55, "Mar": 60})
south = Series({"Feb": 20, "Mar": 25, "Apr": 30})
total = north + south
print(total)
# Jan=NaN (only north), Feb=75, Mar=85, Apr=NaN (only south). Pandas aligns labels.
```

### Q27. Missing label analysis with `isnull`.

```python
d = {"Red": 12, "Blue": 18, "Green": 24}
labels = ["Green", "Red", "Yellow", "Blue"]
result = Series(d, index=labels)
print(result)
print(pd.isnull(result))
# Yellow contains NaN; isnull is True for Yellow. Green, Red and Blue keep numeric values.
```

### Q28. Trace dictionary conversion and reconstruction.

```python
s = Series([90, 75, 82], index=["Ravi", "Aman", "Neha"])
d = s.to_dict()
s2 = Series(d)
print(d)
print(s2)
print(s.equals(s2))
# s2 keeps the labelled data and index order; it is equivalent data but a separate Series object.
```

### Q29. Create, inspect, select, filter, test membership, and convert a product Series.

```python
product_sales = Series([650, 420, 890, 510], index=["Laptop", "Mouse", "Keyboard", "Monitor"])
print("Values:", product_sales.values)
print("Index:", product_sales.index)
print("Keyboard sales:", product_sales["Keyboard"])
print("Sales above 500:")
print(product_sales[product_sales > 500])
print("Mouse exists:", "Mouse" in product_sales.index)
product_dict = product_sales.to_dict()
print(product_dict)
```

### Q30. Full analysis of ordered index, NaN, filter, notnull, and names.

```python
data = {"A": 15, "B": 35, "C": 25}
s = Series(data, index=["C", "A", "D", "B"])
s.name = "Observations"
s.index.name = "Category"
print("Series:")
print(s)
print()
print("Filter > 20:")
print(s[s > 20])
print()
print("Not null:")
print(pd.notnull(s))
print("Series name:", s.name)
print("Index name:", s.index.name)
# Index order: C, A, D, B. D is NaN. Filter returns C=25 and B=35. notnull: C=True, A=True, D=False, B=True.
```
